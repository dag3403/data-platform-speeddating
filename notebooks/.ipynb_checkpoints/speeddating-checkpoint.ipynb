Transformamos variables y revisamos los NAs (Hay que tener en cuenta que las columnas no tienen valores nulos reales None o NaN sino Strings vacíos como "Na" "NaN" o "?" o similares)
La variable que respecta al género la hacemos binaria y cambiamos todos los valores comunes que representan valores faltantes como un valor faltante real, tras esto calculamos el porcentaje de nulos por columna, esto nos ayudará a decidir qué variable eliminar por tener demasiados NA

```python
from pyspark.sql import functions as F

# Convertir género a binario
df = df_raw.withColumn(
    "gender",
    F.when(F.lower(F.trim(F.col("gender"))) == "male", 1)
     .when(F.lower(F.trim(F.col("gender"))) == "female", 0)
     .otherwise(None)
)

# Lista de representaciones comunes de nulos (en texto)
missing_tokens = ["", "NA", "NaN", "nan", "None", "NULL", "null", "?"]

# Reemplazar todos los tokens de nulos por valores nulos reales (None)
for col_name in df.columns:
    df = df.withColumn(
        col_name,
        F.when(F.col(col_name).isin(missing_tokens), None).otherwise(F.col(col_name))
    )

# Verificar porcentaje de valores faltantes después del reemplazo
total = df.count()
exprs = []
for c in df.columns:
    expr = (
        (F.count(F.when(F.col(c).isNull(), c)) / F.lit(total) * 100).alias(c)
    )
    exprs.append(expr)

na_percent_df = df.select(*exprs)
na_percent_pd = na_percent_df.toPandas().T
na_percent_pd.columns = ["porcentaje_NA"]
na_percent_pd = na_percent_pd.sort_values(by="porcentaje_NA", ascending=False)

print("Porcentaje de valores faltantes (tras normalizar nulos):")
print(na_percent_pd.head(25))
```

Como podemos observar la variable expected_num_interested_in_me tiene un 78.51% de datos faltantes por lo que procedemos a eliminarla

Además como se puede observar en la salida de antes la mayoría de variables numéricas son del tipo string lo que para imputar esas variables es necesario pasarlas a una variable del tipo

```python
from pyspark.sql import functions as F
from pyspark.sql.types import FloatType

for c, t in df.dtypes:
    if t == "string":
        df = df.withColumn(c, F.col(c).cast(FloatType()))

# Verificar tipos
df.printSchema()
```

Limpieza de columnas y filas con NA
Vamos a eliminar la columna de expected_num_interested_in_me ya que en su mayoría son datos faltantes e imputar estos valores significa que la mayoría de los valores serán estimados o inventados, lo que puede introducir sesgo y ruido.


```python
# Eliminar columna con demasiados nulos
if "expected_num_interested_in_me" in df.columns:
    df = df.drop("expected_num_interested_in_me")
```

Ahora vamos a ver que todos los valores estén dentro de su rango de valores, para ello hay que tener en cuenta el fichero "speeddating_proyecto_Key" que contiene el rango para cada variable, en este caso vamos a eliminar todas aquellas filas las cuales tenga outliers ya que es mejor para la calidad de los datos que convertirlos en NA

```python
#prueba
from pyspark.sql import functions as F

# ===============================
# Definir rangos permitidos
# ===============================
rangos = {
    "age": (18, 55),
    "wave": (1, 21),
    "gender": (0, 1),
    "age_o": (18, 55),
    "d_age": (0, 37),
    "samerace": (0, 1),
    "importance_same_race": (0, 10),
    "importance_same_religion": (0, 10),
    "pref_o_attractive": (0, 100),
    "pref_o_sincere": (0, 100),
    "pref_o_intelligence": (0, 100),
    "pref_o_funny": (0, 100),
    "pref_o_ambitious": (0, 100),
    "pref_o_shared_interests": (0, 100),
    "attractive_o": (0, 10),
    "sinsere_o": (0, 10),
    "intelligence_o": (0, 10),
    "funny_o": (0, 10),
    "ambitous_o": (0, 10),
    "shared_interests_o": (0, 10),
    "attractive_important": (0, 100),
    "sincere_important": (0, 100),
    "intellicence_important": (0, 100),
    "funny_important": (0, 100),
    "ambtition_important": (0, 100),
    "shared_interests_important": (0, 100),
    "attractive": (0, 10),
    "sincere": (0, 10),
    "intelligence": (0, 10),
    "funny": (0, 10),
    "ambition": (0, 10),
    "attractive_partner": (0, 10),
    "sincere_partner": (0, 10),
    "intelligence_partner": (0, 10),
    "funny_partner": (0, 10),
    "ambition_partner": (0, 10),
    "shared_interests_partner": (0, 10),
    "sports": (0, 10),
    "tvsports": (0, 10),
    "exercise": (0, 10),
    "dining": (0, 10),
    "museums": (0, 10),
    "art": (0, 10),
    "hiking": (0, 10),
    "gaming": (0, 10),
    "clubbing": (0, 10),
    "reading": (0, 10),
    "tv": (0, 10),
    "theater": (0, 10),
    "movies": (0, 10),
    "concerts": (0, 10),
    "music": (0, 10),
    "shopping": (0, 10),
    "yoga": (0, 10),
    "interests_correlate": (-1, 1),
    "expected_happy_with_sd_people": (0, 10),
    "expected_num_matches": (0, 20),
    "like": (0, 10),
    "guess_prob_liked": (0, 10),
    "met": (0, 1),
    "match": (0, 1)
}

# ===============================
# Mostrar cantidad inicial
# ===============================
total_inicial = df.count()
print(f"Filas totales antes de limpiar: {total_inicial}")

# ===============================
# Crear máscara que permita NA
# ===============================
mask = None
fuera_rango = {}

for col, (minv, maxv) in rangos.items():
    if col in df.columns:
        # Considerar NA como válido, solo filtrar si el valor existe y está fuera de rango
        cond = (F.col(col).isNull()) | ((F.col(col) >= minv) & (F.col(col) <= maxv))
        fuera_rango[col] = df.filter(~cond).count()
        if mask is None:
            mask = cond
        else:
            mask = mask & cond

# Filtrar filas con valores fuera de rango (NA se mantienen)
df = df.filter(mask)

# ===============================
# Verificar sumas ≈ 100 para preferencias e importancia
# ===============================
from functools import reduce
from pyspark.sql import functions as F

pref_cols = [
    "pref_o_attractive", "pref_o_sincere", "pref_o_intelligence",
    "pref_o_funny", "pref_o_ambitious", "pref_o_shared_interests"
]

imp_cols = [
    "attractive_important", "sincere_important", "intellicence_important",
    "funny_important", "ambtition_important", "shared_interests_important"
]

for cols in [pref_cols, imp_cols]:
    # Verificar que todas las columnas existen
    if all(c in df.columns for c in cols):
        # Crear columna temporal con la suma
        df = df.withColumn("suma_tmp", sum(F.col(c) for c in cols))
        
        # Crear condición de NA: True si al menos una columna del grupo es NA
        na_cond = reduce(lambda a, b: a | b, [F.col(c).isNull() for c in cols])
        
        # Filtrar filas cuya suma ≈100 o que tengan NA en alguna columna
        df = df.filter((F.abs(F.col("suma_tmp") - 100) <= 1) | na_cond)
        
        # Eliminar columna temporal
        df = df.drop("suma_tmp")


# ===============================
# Mostrar resultados finales
# ===============================
total_final = df.count()

print("\nFilas fuera de rango por columna:")
for k, v in fuera_rango.items():
    print(f"{k}: {v}")

print(f"\nFilas totales antes de limpiar: {total_inicial}")
print(f"Filas totales después de limpiar: {total_final}")

df_clean = df
```

Ahora tras ver los valores faltantes y outliers del dataset tenemos la opción de borrar NAs o sustituirlos por imputación (media, moda, modelos de regresión...). En este caso he decidido borrar los outliers e imputar las observaciones que contengan NAs mediante el método de IterativeImputer el cual modela cada característica con valores faltantes como una función de otras características y utiliza esa estimación para la imputación. En este dataset es una opción muy util ya que existen relaciones lineales entre variables y este método aprobecha estas relaciones

```python
# ===============================
# Importar librerías
# ===============================
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
import pandas as pd
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

# Spark session
spark = SparkSession.builder.getOrCreate()

# ===============================
# Convertir DataFrame PySpark a Pandas
# ===============================
df_pd = df.toPandas()

# ===============================
# Seleccionar columnas numéricas para imputar
# ===============================
# IterativeImputer funciona solo con valores numéricos
num_cols = df_pd.select_dtypes(include=['number']).columns.tolist()
df_num = df_pd[num_cols]

# ===============================
# Inicializar IterativeImputer
# ===============================
imputer = IterativeImputer(max_iter=10, random_state=42)

# ===============================
# Ajustar e imputar
# ===============================
df_imputed_array = imputer.fit_transform(df_num)

# Convertir de nuevo a DataFrame Pandas
df_imputed = pd.DataFrame(df_imputed_array, columns=num_cols)

# ===============================
# Reemplazar las columnas numéricas imputadas en el DataFrame original
# ===============================
df_pd[num_cols] = df_imputed

# ===============================
# Convertir de nuevo a PySpark
# ===============================
df_clean = spark.createDataFrame(df_pd)

# ===============================
# Verificar resultados
# ===============================
print("=== Valores faltantes DESPUÉS de imputar ===")
df_clean.select([F.count(F.when(F.col(c).isNull(), c)).alias(c) for c in df_clean.columns]).show()
```

```python
from pyspark.sql import functions as F

# Contar el total de filas
total_rows = df_clean.count()

# Calcular conteo y porcentaje de valores faltantes por columna
missing_stats = df_clean.select([
    F.count(F.when(F.col(c).isNull() | F.isnan(F.col(c)), c)).alias(c)
    for c in df_clean.columns
])

# Convertir a Pandas para manipulación más fácil
missing_stats_pd = missing_stats.toPandas().T.reset_index()
missing_stats_pd.columns = ['columna', 'faltantes']
missing_stats_pd['porcentaje'] = (missing_stats_pd['faltantes'] / total_rows * 100).round(2)

# Mostrar resultados
print("=== Valores faltantes por columna ===")
print(missing_stats_pd)
```
